# Multiple Linear Regression On California Housing Dataset

```python
# Importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Read the dataset
from sklearn.datasets import fetch_california_housing
california = fetch_california_housing()
```

### Prepare the dataframe

```python
dataset = pd.DataFrame(california.data, columns = california.feature_names)
dataset.head()
```

```python
# Adding price column in dataset
dataset["Price"] = california.target
```

```python
dataset.head()
```

### EDA

```python
# info about dataset
dataset.info()
```

```python
# checking the null values
dataset.isnull().sum()
```

```python
# statistical analysis
dataset.describe()
```

```python
# correlation
dataset.corr()
```

```python
# heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(dataset.corr(), annot=True)
```

### Independent and Dependent Features

```python
X = dataset.drop("Price", axis=1)
y = dataset["Price"]
X.shape, y.shape
```

```python
X[:5]
```

```python
y[:5]
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

### Standardize the data

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train[:5]
```

```python
y_train[:5]
```

### Model Training 


```python
from sklearn.linear_model import LinearRegression
regression = LinearRegression()
regression.fit(X_train, y_train)
```

```python
# coefficients or slopes
regression.coef_
```

```python
# intercept
regression.intercept_
```

```python
# Prediction for the test data
y_pred = regression.predict(X_test)
y_pred[:2]
```

```python
y_test[:2]
```

### Performance Metrics

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mse)
print(mse)
print(mae)
print(rmse)
```

```python
# R square and Adjusted R square
score = r2_score(y_test, y_pred)
adjusted_score = 1 - (1-score)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1)
print(score)
print(adjusted_score)
```

### Assumptions (if prediction is good or not)

```python
# 1. plot Truth Data vs Predicted Data  (expected: linear distribution)
plt.scatter(y_test, y_pred)
plt.xlabel("Test Truth Data")
plt.ylabel("Test Predicted Data")
plt.show()
```

```python
# 2. Plot kde plot on residuals  (expected: normal distribution)
residuals = y_test - y_pred
sns.kdeplot(residuals, color='g')
```

```python
# 3. Plot residuals vs predictions  (expected: uniform distribution)
plt.scatter(y_pred, residuals)
plt.xlabel("Prediction")
plt.ylabel("Residuals")
```

### Pickling

Python pickle module is used for serialising and de-serialising a Python Object structure. Any object in python can be pickled so that it can be saved on disk. What pickle does is that it "serialises" the object first before writing it to file. Pickling is a way to convert a python object (list, dict, etc.) into a character stream. The idea is that this character stream contains all the information necessary to reconstruct the object in another python script.

```python
# Saving a model
import pickle
pickle.dump(regression, open("regressor.pkl", "wb"))
```

```python
# Loading a model
model = pickle.load(open("regressor.pkl", "rb"))
```

```python
model.predict(X_test)
```
